```python
# ============================================================
# Fill Missing Values in Income Column
# Using Median because Income contains numeric values
# ============================================================

median_income = df["Income"].median()

df["Income"] = df["Income"].fillna(median_income)

print("Missing Values After Filling Income :")
print(df["Income"].isnull().sum())
```

```python
orig = pd.read_csv('../dataset/marketing_campaign.csv', sep='\t')
filled_mask = orig["Income"].isna()
df.loc[filled_mask, ["ID", "Income"]]
```

```python
# ============================================================
# Convert Customer Joining Date into Datetime Format
# ============================================================

# Use dayfirst=True because dates are in DD-MM-YYYY format.
# Use errors='coerce' to convert invalid values to NaT instead of raising an error.
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True, errors="coerce")

df["Dt_Customer"]
```

```python
# ============================================================
# Drop Unnecessary Columns
# ============================================================

df = df.drop(columns=["Z_CostContact", "Z_Revenue"])

df.head()
```

```python
# ============================================================
# Create Age Column
# ============================================================

current_year = pd.Timestamp.today().year

df["Age"] = current_year - df["Year_Birth"]

df[["Year_Birth", "Age"]].head()
```

```python
# ============================================================
# Create Total Children Column
# ============================================================

df["Total_Children"] = df["Kidhome"] + df["Teenhome"]

df[["Kidhome", "Teenhome", "Total_Children"]].head()
```

```python
# ============================================================
# Create Total Spending Column
# ============================================================

df["Total_Spending"] = (
    df["MntWines"]
    + df["MntFruits"]
    + df["MntMeatProducts"]
    + df["MntFishProducts"]
    + df["MntSweetProducts"]
    + df["MntGoldProds"]
)

df[["Total_Spending"]].head()
```

```python
# ============================================================
# Create Customer Tenure
# Number of Days Since Customer Joined
# ============================================================

today = pd.Timestamp.today()

df["Customer_Tenure"] = (today - df["Dt_Customer"]).dt.days

df[["Dt_Customer", "Customer_Tenure"]].head()
```

```python
# ============================================================
# Create Income Category
# ============================================================

income_bins = [0, 30000, 70000, df["Income"].max()]

income_labels = ["Low Income", "Medium Income", "High Income"]

df["Income_Category"] = pd.cut(
    df["Income"],
    bins=income_bins,
    labels=income_labels
)

df[["Income", "Income_Category"]].head()
```

```python
# ============================================================
# Create Age Group
# ============================================================

age_bins = [0, 30, 50, 100]

age_labels = ["Young", "Adult", "Senior"]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)

df[["Age", "Age_Group"]].head()
```

```python
# ============================================================
# Create Parent Category
# ============================================================

df["Is_Parent"] = df["Total_Children"] > 0

df[["Total_Children", "Is_Parent"]].head()
```
