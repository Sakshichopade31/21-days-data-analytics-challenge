#### Checking and analyzing null values in Director column

```python
print("Director NULL values:", df['director'].isnull().sum())
print("Percentage:", (df['director'].isnull().sum() / len(df)) * 100, "%")
df[df['director'].isnull()]['type'].value_counts()
```

```python
df[df['director'].notna()] #bas director wale analyze krenge 
```

```python
df['director'].fillna('Unknown', inplace=True)
```

### Analyzing null values of cast

```python

print("Cast NULL values:", df['cast'].isnull().sum())
print("Percentage:", (df['cast'].isnull().sum() / len(df)) * 100, "%")
print("\nNull values by type:")
print(df[df['cast'].isnull()]['type'].value_counts())
```

```python
df_clean = df.dropna(subset=['cast'])
df_clean.shape
```

```python
print("Original data shape:", df.shape)
print("Cleaned data shape:", df_clean.shape)
print("Rows removed:", df.shape[0] - df_clean.shape[0])
print("\n" + "="*50)
print("Cleaned Data:")
print("="*50)
df_clean.head()
```
