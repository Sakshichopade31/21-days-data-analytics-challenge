```python
# Reload the original data and create a clean copy

df = pd.read_csv('../dataset/netflix_titles.csv')
df_clean = df.copy()

# Fill missing values in selected columns
df_clean['director'] = df_clean['director'].fillna('Unknown')
df_clean['cast'] = df_clean['cast'].fillna('Unknown')
df_clean['country'] = df_clean['country'].fillna('Unknown')
df_clean['date_added'] = df_clean['date_added'].fillna('Unknown')
df_clean['duration'] = df_clean['duration'].fillna('Unknown')

# Drop rows where cast is still missing
df_clean = df_clean.dropna(subset=['cast'])

print("Original data shape:", df.shape)
print("Cleaned data shape:", df_clean.shape)
print("Rows removed:", df.shape[0] - df_clean.shape[0])
print("\n" + "="*50)
print("Cleaned Data:")
print("="*50)
df_clean.head(10)
```

###  EDA

```python
# Movie vs TV Show Distribution
plt.figure(figsize = (8, 6))
sns.countplot(data=df_clean, x='type', palette='Set2')
plt.title('Distribution of Movies and TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
```

```python
# Top Country
top_countries = df_clean['country'].value_counts().head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='Set1')
plt.title('Top 10 Countries with Most Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')   
```

```python
## 3. Rating distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=df_clean, x='rating', order=df_clean['rating'].value_counts().index, palette='Set3')
plt.xticks(rotation=45)
plt.title('Rating Distribution')
plt.show()
```

```python
# 4. Release year trend
year_trend = df_clean['release_year'].value_counts().sort_index()
plt.figure(figsize=(10, 4))
year_trend.plot(kind='line', marker='o',color='red')
plt.title('Content Added by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.show()
```

```python
# Top 10 genres
from collections import Counter

genres = df_clean['listed_in'].dropna().str.split(',').explode()
genres = genres.str.strip()
genres_top = genres.value_counts().head(10)

print('Top 10 genres:')
print(genres_top)

# Top 10 directors
directors = df_clean['director'].dropna().str.split(',').explode()
directors = directors.str.strip()
directors_top = directors.value_counts().head(10)

print('\nTop 10 directors:')
print(directors_top)

# Month-wise addition trend
# Keep only rows that are not 'Unknown'
df_month = df_clean[df_clean['date_added'] != 'Unknown'].copy()
df_month['date_added'] = pd.to_datetime(df_month['date_added'], format='mixed', errors='coerce')
df_month = df_month.dropna(subset=['date_added'])
monthly_trend = df_month['date_added'].dt.to_period('M').value_counts().sort_index()

print('\nMonth-wise addition trend:')
print(monthly_trend.head(12))

# Complete analytics summary
summary = {
    'Total Titles': len(df_clean),
    'Movies': (df_clean['type'] == 'Movie').sum(),
    'TV Shows': (df_clean['type'] == 'TV Show').sum(),
    'Top Country': df_clean['country'].value_counts().idxmax(),
    'Top Rating': df_clean['rating'].value_counts().idxmax(),
    'Top Genre': genres_top.index[0],
    'Top Director': directors_top.index[0]
}

print('\nComplete Analytics Summary:')
for k, v in summary.items():
    print(f'{k}: {v}')
```

```python
# 1. Dataset Overview
print('Dataset shape:', df_clean.shape)
print('Columns:', df_clean.columns.tolist())
print('\nSample rows:')
df_clean.head(3)

# 2. Duplicate Records Check
print('Duplicate Rows:', df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()
print('Shape after removing duplicates:', df_clean.shape)

# 3. Missing Values Visualization
plt.figure(figsize=(8, 5))
sns.heatmap(df_clean.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()

# 4. Movies vs TV Shows Percentage
print(df_clean['type'].value_counts(normalize=True) * 100)

plt.figure(figsize=(6, 6))
df_clean['type'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='Set2')
plt.title('Movies vs TV Shows Distribution')
plt.ylabel('')
plt.show()

# 5. Top 10 Directors
plt.figure(figsize=(10, 5))
directors_top = df_clean['director'].value_counts().head(10)
sns.barplot(x=directors_top.values, y=directors_top.index, palette='magma')
plt.title('Top 10 Directors')
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.show()

# 6. Top 10 Actors
actors = df_clean['cast'].dropna().str.split(',').explode().str.strip()
actors_top = actors.value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=actors_top.values, y=actors_top.index, palette='viridis')
plt.title('Top 10 Actors')
plt.xlabel('Number of Titles')
plt.ylabel('Actor')
plt.show()

# 7. Top Genres
genres = df_clean['listed_in'].dropna().str.split(',').explode().str.strip()
genres_top = genres.value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=genres_top.values, y=genres_top.index, palette='cividis')
plt.title('Top 10 Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()

# 8. Movies Released Per Decade
decade_bins = [1990, 2000, 2010, 2020, 2030]
decade_labels = ['1990s', '2000s', '2010s', '2020s']
df_clean['decade'] = pd.cut(df_clean['release_year'], bins=decade_bins, labels=decade_labels, right=False)

plt.figure(figsize=(8, 5))
df_clean['decade'].value_counts().reindex(decade_labels).plot(kind='bar', color='steelblue')
plt.title('Movies Released Per Decade')
plt.xlabel('Decade')
plt.ylabel('Number of Titles')
plt.show()

# 9. Content Added Over Time
monthly_trend = df_clean[df_clean['date_added'] != 'Unknown'].copy()
monthly_trend['date_added'] = pd.to_datetime(monthly_trend['date_added'], format='mixed', errors='coerce')
monthly_trend = monthly_trend.dropna(subset=['date_added'])
monthly_counts = monthly_trend['date_added'].dt.to_period('M').value_counts().sort_index()
year_counts = monthly_trend['date_added'].dt.to_period('Y').value_counts().sort_index()

plt.figure(figsize=(12, 4))
monthly_counts.plot(kind='line', marker='o', color='purple')
plt.title('Month-wise Additions')
plt.xlabel('Month')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(10, 4))
year_counts.plot(kind='line', marker='o', color='green')
plt.title('Year-wise Additions')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()

# 10. Country-wise Movie Distribution
country_top = df_clean['country'].value_counts().head(15)
plt.figure(figsize=(10, 5))
sns.barplot(x=country_top.values, y=country_top.index, palette='pastel')
plt.title('Top 15 Countries by Content')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()

# 11. Ratings Analysis
print('Most common rating:', df_clean['rating'].value_counts().idxmax())
print('Kids content count:', df_clean[df_clean['rating'].isin(['TV-Y', 'TV-Y7', 'G', 'PG', 'PG-13'])].shape[0])
print('Adult content count:', df_clean[df_clean['rating'].isin(['TV-MA', 'R', 'NC-17'])].shape[0])

plt.figure(figsize=(10, 5))
sns.countplot(data=df_clean, x='rating', order=df_clean['rating'].value_counts().index, palette='Set2')
plt.xticks(rotation=45)
plt.title('Rating Distribution')
plt.show()

# 12. Movie Duration Analysis
movie_duration = df_clean[df_clean['type'] == 'Movie'].copy()
movie_duration['duration_min'] = movie_duration['duration'].str.extract(r'(\d+)').astype(float)

plt.figure(figsize=(8, 5))
sns.histplot(movie_duration['duration_min'].dropna(), bins=20, color='orange')
plt.title('Movie Duration Distribution')
plt.xlabel('Duration (minutes)')
plt.ylabel('Count')
plt.show()

# 13. TV Show Seasons Analysis
tv_seasons = df_clean[df_clean['type'] == 'TV Show'].copy()
tv_seasons['season_count'] = tv_seasons['duration'].str.extract(r'(\d+)').astype(float)

plt.figure(figsize=(8, 5))
sns.histplot(tv_seasons['season_count'].dropna(), bins=10, color='brown')
plt.title('TV Show Season Count Distribution')
plt.xlabel('Number of Seasons')
plt.ylabel('Count')
plt.show()

# 14. Oldest & Latest Movie
print('Oldest release year:', df_clean['release_year'].min())
print('Latest release year:', df_clean['release_year'].max())

# 15. Longest Movie
longest_movie = movie_duration.sort_values('duration_min', ascending=False).head(1)
print('Longest movie:', longest_movie[['title', 'duration_min']].iloc[0].to_dict())

# 16. Netflix Content Growth
year_growth = df_clean['release_year'].value_counts().sort_index()
plt.figure(figsize=(10, 4))
year_growth.plot(kind='line', marker='o', color='royalblue')
plt.title('Netflix Content Growth by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()

# 17. Insights Section
print('Insight 1: Movies significantly outnumber TV shows in the dataset.')
print('Insight 2: The United States contributes the highest number of titles.')
print('Insight 3: TV-MA is the most common rating among titles.')
print('Insight 4: Drama and International Movies dominate the catalog.')
print('Insight 5: Netflix content expanded rapidly after 2015.')

# 18. Conclusion
print('\nConclusion')
print(f'• Dataset contains {len(df_clean)} titles.')
print('• Movies dominate the Netflix catalog.')
print('• United States contributes the maximum number of titles.')
print('• TV-MA is the most frequent rating.')
print('• Drama is one of the most popular genres.')
print('• Netflix content expanded rapidly after 2015.')

# 19. Business Recommendations
print('\nBusiness Recommendations')
print('• Increase kids-oriented content in underrepresented regions.')
print('• Invest more in high-performing genres such as dramas and international content.')
print('• Focus on countries showing growing demand for local-language productions.')
print('• Continue producing TV-MA content because it dominates the catalog.')
```
